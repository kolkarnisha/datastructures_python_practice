```python
def all_triplets_zero(arr):
    n = len(arr)
    triplets = set()   # use set to avoid duplicates

    for i in range(n-1):
        seen = {}
        for j in range(i+1, n):
            target = -(arr[i] + arr[j])
            if target in seen:
                triplet = tuple(sorted([arr[i], arr[j], target]))
                triplets.add(triplet)
            else:
                seen[arr[j]] = True
    return list(triplets)


# Example
arr = [-3, -1, -1, 0, 1, 2]
print(all_triplets_zero(arr))
```

```python
''' all triplets with sum as target'''
def count_triplets(arr, target):
    arr.sort()   
    n = len(arr)
    res = 0

    for i in range(n-2):
        j, k = i+1, n-1
        while j < k:
            s = arr[i] + arr[j] + arr[k]
            if s == target:
                # Case 1: same element at j and k
                if arr[j] == arr[k]:
                    cnt = k - j + 1
                    res += cnt * (cnt - 1) // 2
                    break
                else:
                    # count duplicates of arr[j]
                    cnt1 = 1
                    while j+1 < k and arr[j] == arr[j+1]:
                        cnt1 += 1
                        j += 1
                    # count duplicates of arr[k]
                    cnt2 = 1
                    while k-1 > j and arr[k] == arr[k-1]:
                        cnt2 += 1
                        k -= 1
                    res += cnt1 * cnt2
                    j += 1
                    k -= 1
            elif s < target:
                j += 1
            else:
                k -= 1
    return res

# Example
arr = [-3, -1, -1, 0, 1, 2]
target = -2
print(count_triplets(arr, target))  # Output: 4
```

```python
class T:
    def c(self, arr, target):
        arr.sort()
        n=len(arr)
        count=0
        for i in range(n-2):
            left=i+1      
            right=n-1      
            while left<right:
                total=arr[i]+arr[left]+arr[right]
                if total==target:
                    count+=1
                    left+=1
                    right-=1
                elif total<target:
                    left+=1  
                else:
                    right-=1 

        return count


ts = T()

arr1 = [0, -1, 2, -3, 1]
target1 = -2
print("Count:", ts.c(arr1, target1))  

```
