```python
class Solution(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        seen = {}
        for i, num in enumerate(nums):
            complement = target - num
            if complement in seen:
                return [seen[complement], i]
            seen[num] = i
```

```python
# next permutation
def nextper(a):
    i = len(a) - 2
    while i >= 0 and a[i] >= a[i + 1]:
        i -= 1

    if i >= 0:
        j = len(a) - 1
        while a[j] <= a[i]:
            j -= 1
        a[i], a[j] = a[j], a[i]

    a[i + 1:] = reversed(a[i + 1:])
    return a

print(nextper([10, 60, 70, 40, 50, 90]))
```

```python
a = [2, 3, -8, 7, -1, 2, 3]
# Kadane's algorithm: find the maximum sum of a contiguous subarray

current_sum = a[0]
maximum_sum = a[0]
start = 0
best_start = 0
best_end = 0

for i in range(1, len(a)):
    if a[i] > current_sum + a[i]:
        current_sum = a[i]
        start = i
    else:
        current_sum += a[i]

    if current_sum > maximum_sum:
        maximum_sum = current_sum
        best_start = start
        best_end = i
    print(maximum_sum)

print("Maximum sum:", maximum_sum)
print("Subarray:", a[best_start:best_end + 1])
```

```python
arr = [8, -8, 9, -9, 10, -11, 18]


def kadane(numbers):
    current = best = numbers[0]

    for number in numbers[1:]:
        current = max(number, current + number)
        best = max(best, current)

    return best


def circular_max_subarray(numbers):
    maximum_sum = kadane(numbers)

    # If every number is negative, the non-circular result is correct.
    if maximum_sum < 0:
        return maximum_sum

    total = sum(numbers)
    minimum_sum = -kadane([-number for number in numbers])
    return max(maximum_sum, total - minimum_sum)


def circular_min_subarray(numbers):
    minimum_sum = -kadane([-number for number in numbers])

    # If every number is positive, the non-circular result is correct.
    if minimum_sum > 0:
        return minimum_sum

    total = sum(numbers)
    maximum_sum = kadane(numbers)
    return min(minimum_sum, total - maximum_sum)


print("Maximum circular subarray sum:", circular_max_subarray(arr))
print("Minimum circular subarray sum:", circular_min_subarray(arr))
```
