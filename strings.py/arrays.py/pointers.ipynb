```python
class S:
    def c(self, arr, target):
        res = 0
        n = len(arr)
        for i in range(0, n-2):
            j = i + 1
            k = n - 1
            while j < k:
                s = arr[i] + arr[j] + arr[k]
                if s < target:
                    j += 1
                elif s > target:
                    k -= 1   
                else:
                    elel = arr[j]
                    ele2 = arr[k]
                    cnt1 = cnt2 = 0
                    while j <= k and arr[j] == elel:
                        j += 1
                        cnt1 += 1
                    while j <= k and arr[k] == ele2:
                        k -= 1
                        cnt2 += 1
                    if elel == ele2:
                        res += cnt1 * (cnt1 - 1) // 2
                    else:
                        res += cnt1 * cnt2
        return res

arr = [-3, -1, -1, 0, 1, 2]
target = -2
o = S()
print(o.c(arr, target))   
```

```python
'''pairs with limited sum '''
''' eey two elements sum less than taget
arr=[7,2,5,3] target=8   output=2
     |
     first sort 
     2  3  5  7
     |         |
     left      '''
arr=[7,2,5,3] 
target=8
arr.sort()
res=0
left=0
right=len(arr)-1
while left<=right:
    sum=arr[left]+arr[right]
    if sum<target:
        res+=right-left
        left+=1
    else:
        right-=1
print(res)
```
