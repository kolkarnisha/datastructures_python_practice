```python
class BinaryAdder:
    def trimtosize(self, s):
        first = s.find("i")
        return s[first:] if first != -1 else "0"

    def addbinary(self, s1, s2):
        # optional trim step if you want to preprocess
        s1 = self.trimtosize(s1) if "i" in s1 else s1
        s2 = self.trimtosize(s2) if "i" in s2 else s2

        n, m = len(s1), len(s2)
        if n < m:
            s1, s2 = s2, s1
            n, m = m, n

        j = m - 1
        carry = 0
        result = []

        for i in range(n - 1, -1, -1):
            bit1 = int(s1[i])
            bitsum = bit1 + carry
            if j >= 0:
                bit2 = int(s2[j])
                bitsum += bit2
                j -= 1
            bit = bitsum % 2
            carry = bitsum // 2
            result.append(str(bit))

        if carry > 0:
            result.append(str(carry))

        return "".join(result[::-1])


# Usage
adder = BinaryAdder()
print(adder.addbinary("1101", "111"))  # 10100
```

```python
# KMP algorithm (Knuth-Morris-Pratt)
class Solution:
    def construct(self, pat):
        lps = [0] * len(pat)
        i = 1
        j = 0

        while i < len(pat):
            if pat[i] == pat[j]:
                j += 1
                lps[i] = j
                i += 1
            elif j != 0:
                j = lps[j - 1]
            else:
                i += 1

        return lps

    def search(self, pat, txt):
        if pat == "":
            return []

        lps = self.construct(pat)
        result = []
        i = j = 0

        while i < len(txt):
            if txt[i] == pat[j]:
                i += 1
                j += 1
                if j == len(pat):
                    result.append(i - j)
                    j = lps[j - 1]
            elif j != 0:
                j = lps[j - 1]
            else:
                i += 1

        return result

obj = Solution()
print(obj.search("abc", "ababcababcabc"))  # [2, 7, 10]
```
