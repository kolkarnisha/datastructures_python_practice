```python
''' search in sorted marix
input mat=[[1,5,9],[14,20,21],[30,34,43]]
x=14
first flat them rather than matrix
then use binary
'''
class s:
    def bin(self,mat,x):
        n=len(mat)
        m=len(mat[0])
        low=0
        high=n*m-1
        while low<=high:
            mid=(low+high)//2
            row=mid//m
            col=mid%m
            if x==mat[row][col]:
                return True
            elif (x<mat[row][col]):
                high=mid-1
            else:
                low=mid+1
        return False
mat=[[1,5,9],[14,20,21],[30,34,43]]
x=14
o=s()
o.bin(mat,x)


```
