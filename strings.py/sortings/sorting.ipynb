```python
''' allocate minimum pages
one library has books
first book has 12 pages 2nd has 34 ...
all books should be allocated
rules one book should not allocate to two should
sequence order of allocation 
no same book allocation for two student
when allocation minimum no.of books to be allocated to students
input arr=[12,34,67,90], k=2   2students
outut=113
minimum pages
[12] and [34,67,90]   maxium 191
[12,34] and [67,90]      max 157
'''
class sol:
    def check(self,arr,k,pagelimit):
        cnt=1
        pagesum=0
        for i in arr:
            if pagesum+i>pagelimit:
                cnt+=1
                pagesum=i
            else:
                pagesum+=i
        return cnt<=k
    def allocate(self,arr,k):
        res=-1
        if k>len(arr):
            return -1
        low=max(arr)
        high=sum(arr)
        while low<=high:
            mid=(low+high)//2
            if self.check(arr,k,mid):
                res=mid
                high=mid-1
            else:
                low=mid+1
        return res
arr=[12,34,67,90]
k=2
obj=sol()
obj.allocate(arr,k)
```
